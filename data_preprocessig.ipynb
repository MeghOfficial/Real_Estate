```python
# Function to convert all prices values into Lakhs

def convert_to_lakh(price):

    # Remove ₹ symbol and extra spaces
    price = price.replace("₹", "").strip()

    # Convert Crore to Lakh
    if "Cr" in price:
        value = float(price.replace("Cr", "").strip()) * 100

    # Keep Lakh as it is
    elif "Lac" in price:
        value = float(price.replace("Lac", "").strip())

    else:
        return None

    return value
```

# Feature Engineering

 We can see that Description column contains a lot of unique values because it is a combination of multiple text fields. AS there are not common pattern so Regex or any other method cannot be applied to extract specific information from Description column. 

 So i decided to use baseclass schema and LLM to extract specific information from Description column and create new columns for that information. This will help in better analysis and modeling.

```python
from groq import Groq
from pydantic import BaseModel, Field, field_validator
from typing import List, Optional
import json
from dotenv import load_dotenv
import os

# Load .env file
load_dotenv()

# Read API key from .env
groq_api_key = os.getenv("GROQ_API_KEY")

# Create Groq client
client = Groq(api_key=groq_api_key)


# Pydantic Schema
class PropertyDetails(BaseModel):

    # Basic Property Info
    bedrooms: Optional[int] = None
    bathrooms: Optional[int] = None
    balconies: Optional[int] = None

    # Area / Floor
    current_floor: Optional[str] = None
    total_floors: Optional[int] = None
    total_floors_house: Optional[int] = None

    # Property Characteristics
    facing: Optional[str] = None
    furnishing_status: Optional[str] = None
    property_type: Optional[str] = None

    # Combined Lists - allow optional strings and filter None values
    age_possession: List[Optional[str]] = Field(default_factory=list)
    amenities: List[Optional[str]] = Field(default_factory=list)
    additional_rooms: List[Optional[str]] = Field(default_factory=list)
    nearby: List[Optional[str]] = Field(default_factory=list)
    additional_insights: List[Optional[str]] = Field(default_factory=list)

    @field_validator('age_possession', 'amenities', 'additional_rooms', 'nearby', 'additional_insights', mode='before')
    @classmethod
    def filter_none_values(cls, v):
        """Remove None values from lists"""
        if isinstance(v, list):
            return [item for item in v if item is not None]
        return v


# LLM Extraction Function
def extract_with_llm(description):

    prompt = f"""
    You are an expert real estate information extraction system.

    Extract structured property details from the property description.

    Rules:
    - Return ONLY valid JSON.
    - Missing values should be null.
    - Lists must always return arrays [] even if empty.
    - Normalize values wherever possible.

    Fields to extract:

    1. bedrooms -> integer
    2. bathrooms -> integer
    3. balconies -> integer

    4. current_floor -> string
    5. total_floors -> integer
    6. total_floors_house -> integer

    7. facing -> string
    8. furnishing_status -> furnished / semi-furnished / unfurnished
    9. property_type -> apartment / villa / bungalow / penthouse / etc.

    10. age_possession -> list containing:
        - property age
        - possession details
        Example:
        ["0-1 year old", "ready to move"]

    11. amenities -> list containing amenities like:
        - lift
        - garden
        - parking
        - clubhouse
        - swimming pool
        - gym
        - security
        - power backup
        etc.

    12. additional_rooms -> list containing:
        - pooja room
        - servant room
        - study room
        - store room
        - terrace
        - utility room
        etc.

    13. nearby -> list containing nearby landmarks:
        - school
        - hospital
        - metro
        - mall
        - highway
        etc.

    14. additional_insights -> list containing:
        - vastu compliant
        - corner property
        - luxury property
        - road facing
        - garden facing
        - park facing
        etc.

    Property Description:
    \"\"\"{description}\"\"\"
    """

    response = client.chat.completions.create(

        # model of LLM to use for extraction
        model="llama-3.1-8b-instant",

        messages=[
            {
                "role": "system",
                "content": "You are a highly accurate real estate data extraction engine."
            },
            {
                "role": "user",
                "content": prompt
            }
        ],

        response_format={"type": "json_object"},
        temperature=0

    )

    extracted_json = json.loads(
        response.choices[0].message.content
    )

    # Validate output
    validated_output = PropertyDetails(**extracted_json)

    return validated_output
```

```python
# test the function with a sample description

description = """
Luxury 3 BHK semi-furnished apartment with 2 bathrooms and 3 balconies.
Includes lift, clubhouse, swimming pool, gym, and security.
East facing and vastu compliant.
Ready to move property near metro station and hospital.
"""

result = extract_with_llm(description)

print(result.model_dump())
```

```python
# Apply LLM Extraction to All Descriptions

# Initialize new columns for extracted features (if not already done)
if 'bedrooms' not in df.columns:
    df['bedrooms'] = None
    df['bathrooms'] = None
    df['balconies'] = None
    df['current_floor'] = None
    df['total_floors'] = None
    df['total_floors_house'] = None
    df['facing'] = None
    df['furnishing_status'] = None
    df['property_type'] = None
    df['age_possession'] = None
    df['amenities'] = None
    df['additional_rooms'] = None
    df['nearby'] = None
    df['additional_insights'] = None


success_count = 0
error_count = 0
skip_count = 0
start_row = 0

print(f"Starting extraction from row {start_row}")
print(f"Total rows: {len(df) - start_row}")

for idx in range(start_row, len(df)):

    description = df.at[idx, 'Description']

    # Skip empty descriptions
    if pd.isna(description) or str(description).strip() == "":

        skip_count += 1

        print(f"Row {idx} → SKIPPED")

        continue

    try:

        result = extract_with_llm(description)

        # Store extracted features
        df.at[idx, 'bedrooms'] = result.bedrooms
        df.at[idx, 'bathrooms'] = result.bathrooms
        df.at[idx, 'balconies'] = result.balconies
        df.at[idx, 'current_floor'] = result.current_floor
        df.at[idx, 'total_floors'] = result.total_floors
        df.at[idx, 'total_floors_house'] = result.total_floors_house
        df.at[idx, 'facing'] = result.facing
        df.at[idx, 'furnishing_status'] = result.furnishing_status
        df.at[idx, 'property_type'] = result.property_type
        df.at[idx, 'age_possession'] = result.age_possession
        df.at[idx, 'amenities'] = result.amenities
        df.at[idx, 'additional_rooms'] = result.additional_rooms
        df.at[idx, 'nearby'] = result.nearby
        df.at[idx, 'additional_insights'] = result.additional_insights

        success_count += 1

        print(f"Row {idx} → SUCCESS")

    except Exception as e:

        error_count += 1

        print(f"Row {idx} → ERROR : {str(e)}")

        continue


# FINAL SUMMARY
print("\n================ FINAL REPORT ================")

print(f"Total Rows       : {len(df) - start_row}")
print(f"Successful Rows  : {success_count}")
print(f"Error Rows       : {error_count}")
print(f"Skipped Rows     : {skip_count}")

print("=============================================")
```

```python
# Bathroom per bedroom ratio by location based mapped area

ratio_df = (
    df.dropna(subset=["bedrooms", "bathrooms"])
      .groupby("Mapped_Area")[["bedrooms", "bathrooms"]]
      .median()
)

# Create ratio
ratio_df["ratio"] = (
    ratio_df["bathrooms"] / ratio_df["bedrooms"]
)

print(ratio_df.head())
```

```python
# Fill missing bathrooms using mapped area ratio

mask = (
    df["bathrooms"].isna() &
    df["bedrooms"].notna()
)

for idx in df[mask].index:

    mapped_area = df.loc[idx, "Mapped_Area"]
    bedrooms = df.loc[idx, "bedrooms"]

    # Convert to numeric safely
    bedrooms = float(bedrooms)

    if mapped_area in ratio_df.index:

        ratio = ratio_df.loc[mapped_area, "ratio"]

        estimated_bathrooms = round(bedrooms * ratio)

        # Minimum 1 bathroom
        estimated_bathrooms = max(1, estimated_bathrooms)

    else:
        # fallback heuristic
        estimated_bathrooms = max(1, round(bedrooms))

    df.loc[idx, "bathrooms"] = estimated_bathrooms

    print(
        f"Row {idx} | Area: {mapped_area} | "
        f"Bedrooms: {bedrooms} -> Estimated Bathrooms: {estimated_bathrooms}"
    )
```

```python
def extract_age(text):
    if pd.isna(text):
        return "Unknown"
    
    text = str(text).lower()

    if any(x in text for x in ["0-1", "0–1", "1 year", "1-year", "nearly new"]):
        return "New"
    
    if any(x in text for x in ["1-5", "2-5", "1–5"]):
        return "1-5 years"
    
    if any(x in text for x in ["5-10", "5–10"]):
        return "5-10 years"
    
    if any(x in text for x in ["10-20", "10–15", "11-20", "10 year", "10+"]):
        return "10-20 years"
    
    if any(x in text for x in ["20"]):
        return "20+ years"
    
    return "Unknown"
```

```python
def extract_possession(text):
    if pd.isna(text):
        return "Unknown"
    
    text = str(text).lower()

    if "ready" in text or "immediate" in text:
        return "Ready to move"
    
    if "under" in text:
        return "Under construction"
    
    if "vacant" in text:
        return "Vacant"
    
    if "available" in text:
        return "Future possession"
    
    return "Unknown"
```

```python
def clean_age_possession(x):

    # handle missing values
    if x is None:
        return pd.Series(["Unknown", "Unknown"])

    # convert list → string
    x = str(x).lower()

    # handle empty cases
    if x in ["[]", "[null, null]"]:
        return pd.Series(["Unknown", "Unknown"])

    x = x.strip("[]")

    parts = [p.strip() for p in x.split(",")]

    age = parts[0] if len(parts) > 0 else ""
    possession = parts[1] if len(parts) > 1 else ""

    return pd.Series([
        extract_age(age),
        extract_possession(possession)
    ])
```

```python
def normalize_nearby(x):

    if x is None:
        return []

    if isinstance(x, float) and pd.isna(x):
        return []

    if isinstance(x, (list, np.ndarray)):
        nearby = x
    else:
        try:
            nearby = ast.literal_eval(str(x))
        except:
            return []

    cleaned = []

    for item in nearby:

        item = str(item).strip().lower()

        item = nearby_mapping.get(item, item)

        cleaned.append(item)

    return list(set(cleaned))


df["normalized_nearby"] = df["nearby"].apply(
    normalize_nearby
)
```

```python
def calculate_location_score(nearby_list):

    score = 0

    for item in nearby_list:

        score += location_weights.get(item, 0)

    return score


df["location_advantage_score"] = df[
    "normalized_nearby"
].apply(calculate_location_score)
```

```python
# BONUS WEIGHTS FROM additional_insights

insight_bonus = {

    "luxury property": 5,
    "prime location": 4,
    "riverfront": 6,

    "park facing": 3,
    "garden facing": 3,

    "corner property": 2,
    "road facing": 2,

    "vastu compliant": 1,
    "vaastu compliant": 1,

    "newly built": 2,
    "new construction": 2,
    "fresh construction": 2,

    "gated society": 2,

    "under construction": -1,
    "still under construction": -1,
}


# ADD BONUS TO EXISTING luxury_score
def add_insight_bonus(row):

    current_score = row["luxury_score"]

    insights = row["additional_insights"]

    # Handle null
    if insights is None:
        return current_score

    if isinstance(insights, float) and pd.isna(insights):
        return current_score

    # Convert string list to python list
    if not isinstance(insights, list):

        try:
            insights = ast.literal_eval(str(insights))
        except:
            return current_score

    bonus = 0

    for item in insights:

        item = str(item).strip().lower()

        bonus += insight_bonus.get(item, 0)

    return current_score + bonus


# UPDATE luxury_score
df["luxury_score"] = df.apply(
    add_insight_bonus,
    axis=1
)
```
