```python
mask = df['property_type'].eq('other')

df.loc[mask & df['Name'].str.contains('plot', case=False, na=False), 'property_type'] = 'plot'

df.loc[mask & df['Name'].str.contains('flat|apartment', case=False, na=False), 'property_type'] = 'apartment'

df.loc[mask & df['Name'].str.contains('villa', case=False, na=False), 'property_type'] = 'villa'

df.loc[mask & df['Name'].str.contains('house', case=False, na=False), 'property_type'] = 'house'

df.loc[mask & df['Name'].str.contains('builder floor', case=False, na=False), 'property_type'] = 'builder floor'

df.loc[mask & df['Name'].str.contains('penthouse', case=False, na=False), 'property_type'] = 'penthouse'

df.loc[mask & df['Name'].str.contains('farmhouse', case=False, na=False), 'property_type'] = 'farmhouse'
```

```python
mask = df['property_type'].isna()

df.loc[mask & df['Name'].str.contains('plot', case=False, na=False), 'property_type'] = 'plot'

df.loc[mask & df['Name'].str.contains('flat|apartment', case=False, na=False), 'property_type'] = 'apartment'

df.loc[mask & df['Name'].str.contains('villa', case=False, na=False), 'property_type'] = 'villa'

df.loc[mask & df['Name'].str.contains('house', case=False, na=False), 'property_type'] = 'house'

df.loc[mask & df['Name'].str.contains('builder floor', case=False, na=False), 'property_type'] = 'builder floor'

df.loc[mask & df['Name'].str.contains('penthouse', case=False, na=False), 'property_type'] = 'penthouse'

df.loc[mask & df['Name'].str.contains('farmhouse', case=False, na=False), 'property_type'] = 'farmhouse'
```

```python
# Manual verification from 99acres
df.loc[331, 'property_type'] = 'house'      # verified residential house
df.loc[1174, 'property_type'] = 'plot'      # verified residential plot
df.loc[1729, 'property_type'] = 'plot'      # verified large residential plot

# commercial property (shop)
df = df.drop(index=1585)
```

```python
df['property_type'].value_counts()
```

```python
# Apartments
apartment_df = df[
    df['property_type'].isin(
        ['apartment', 'builder floor', 'penthouse']
    )
].copy()

# Houses & Villas
house_villa_df = df[
    df['property_type'].isin(
        ['house', 'villa', 'luxuryhouse']
    )
].copy()

# Plots
plot_df = df[
    df['property_type'].isin(
        ['plot', 'farmhouse']
    )
].copy()
```

```python
print("Apartment:", apartment_df.shape)
print("House/Villa:", house_villa_df.shape)
print("Plot:", plot_df.shape)
```

```python
apartment_df.to_csv("gandhinagar_property_apartments.csv", index=False)
house_villa_df.to_csv("gandhinagar_property_houses.csv", index=False)
plot_df.to_csv("gandhinagar_property_plots.csv", index=False)
```
